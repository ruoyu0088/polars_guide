## P-019

对收据明细数据（df_receipt），按每条记录的销售金额（amount）从高到低添加排名，从开头显示 10 条。显示项目为顾客 ID（customer_id）、销售金额（amount）和添加的排名。销售金额（amount）相同时赋予相同名次。

```python
(
df_receipt
.select(
    pl.col('customer_id', 'amount')
        .sort_by('amount', descending=True)
)
.with_columns(
    pl.col.amount
        .rank(method='min', descending=True)
        .alias('rank')
)
.head(10)
)
```

## P-020

对收据明细数据（df_receipt），按每条记录的销售金额（amount）从高到低添加排名，从开头显示 10 条。显示项目为顾客 ID（customer_id）、销售金额（amount）和添加的排名。销售金额（amount）相同时也要赋予不同名次。

```python
(
df_receipt
.select(
    pl.col('customer_id', 'amount')
        .sort_by('amount', descending=True)
)
.with_columns(
    pl.col.amount
        .rank(method='ordinal', descending=True)
        .alias('rank')
)
.head(10)
)
```
