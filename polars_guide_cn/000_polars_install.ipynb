# 安装与学习资料

## 安装

Polars 的安装本身并不复杂，难点在于选哪一套、与哪些库搭配。这一节先给出安装命令，再说明本书用到的可选依赖各自解决什么问题，最后列出值得参考的资料。截至 2026 年 10 月，最新的稳定版本是 2.0 系列。

```bash
pip install polars
```

如果使用 `conda`，则从 conda-forge 安装。

```bash
conda install -c conda-forge polars
```

本书的代码都在 2.0 上验证过，1.x 的写法有些已经不再适用。Polars 2.0 起 `LazyFrame` 的 `collect()` 默认走流式引擎，内存放不下的数据也能分块处理，代价是 `join`、`group_by` 这类操作不再保证行序。1.x 的代码有些会因此直接出错，也有些只是结果顺序发生变化；迁移时请参考官方的迁移指南。

https://docs.pola.rs/releases/upgrade/2/

### 可选依赖

本书只在需要的地方用到下面这些库，它们都不是 Polars 的必需依赖，请在开始对应章节之前装好。其中 `hvplot` 负责画交互图，`pyarrow` 负责 Arrow 格式互转与观察内部数据结构，`fastexcel` 让 Excel 读得更快，`xlsxwriter` 负责写 Excel。各库的安装命令见下表，装好之后不必再改环境变量。

| 库 | 用途 | 安装 |
|---|---|---|
| `hvplot` / `geoviews` | 用 `.hvplot()` 画图 | `conda install -c conda-forge hvplot geoviews altair` |
| `pyarrow` | 与 Arrow 互转、观察内部数据结构 | `conda install -c conda-forge pyarrow` |
| `fastexcel` | 快速读取 Excel | `conda install -c conda-forge fastexcel` |
| `xlsxwriter` | 写出 Excel | `conda install -c conda-forge xlsxwriter` |

## 学习资料

官方文档是最好的参考。想先建立直觉，可以读 Modern Polars 与两本实战书；想直接找现成的写法，Polars Cookbook 与 Awesome Polars 这两份清单更直接。

* [官方文档](https://docs.pola.rs/)
* [Polars 2.0 迁移指南](https://docs.pola.rs/releases/upgrade/2/)
* [Modern Polars](https://kevinheavey.github.io/modern-polars/)
* [Python Polars: The Definitive Guide](https://github.com/jeroenjanssens/python-polars-the-definitive-guide)
* [Data Analysis with Polars](https://github.com/PacktPublishing/Data-Analysis-with-Polars)
* [Polars Cookbook](https://github.com/escobar-west/polars-cookbook)
* [Awesome Polars](https://github.com/ddotta/awesome-polars)
