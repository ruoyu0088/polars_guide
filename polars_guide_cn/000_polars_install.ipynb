# 安装与学习资料

## 安装

Polars 的安装本身很简单，麻烦的是“装哪一套、跟哪些库搭配”。这一节先给出安装命令，再说明本书用到的可选依赖各自解决什么问题，最后列出值得跟着读的资料。

截至 2026 年 10 月，最新版是 2.0 系列，本书的代码都在 2.0 上验证过。

```bash
pip install polars
```

用 `conda` 的话，从 conda-forge 安装。

```bash
conda install -c conda-forge polars
```

Polars 2.0 起，`LazyFrame` 的 `collect()` 默认走流式引擎：内存放不下的数据也能分块算完，但代价是 `join`、`group_by` 这类操作不再保证行序。1.x 的代码有些会因为这一点直接出错或者结果顺序变掉，迁移时请参考官方的迁移指南。

https://docs.pola.rs/releases/upgrade/2/

### 可选依赖

本书只在需要的地方用到下面这些库，它们都不是 Polars 的必需依赖：`hvplot` 负责画交互图，`pyarrow` 负责 Arrow 格式互转和观察内部数据结构，`fastexcel` 让 Excel 读得更快，`xlsxwriter` 负责写 Excel。请在对应章节之前装好。

| 库 | 用途 | 安装 |
|---|---|---|
| `hvplot` / `geoviews` | 画图（`.hvplot()`） | `conda install -c conda-forge hvplot geoviews altair` |
| `pyarrow` | 与 Arrow 互转、观察内部数据结构 | `conda install -c conda-forge pyarrow` |
| `fastexcel` | 快速读取 Excel | `conda install -c conda-forge fastexcel` |
| `xlsxwriter` | 写出 Excel | `conda install -c conda-forge xlsxwriter` |
