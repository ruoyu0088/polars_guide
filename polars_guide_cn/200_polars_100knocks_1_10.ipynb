## P-005

从收据明细数据(`df_receipt`)中，按销售日(`sales_ymd`)、顾客 ID(`customer_id`)、商品代码(`product_cd`)、销售金额(`amount`)的顺序指定列，抽取满足下列全部条件的数据。

- 顾客 ID(`customer_id`)为 "CS018205000001"
- 销售金额(`amount`)在 1,000 以上

````python
(
df_receipt
.filter(
    (pl.col.customer_id == 'CS018205000001')
    & (pl.col.amount >= 1000)
)
.select('sales_ymd', 'customer_id', 'product_cd', 'amount')
)
````

## P-006

从收据明细数据(`df_receipt`)中，按销售日(`sales_ymd`)、顾客 ID(`customer_id`)、商品代码(`product_cd`)、销售数量(`quantity`)、销售金额(`amount`)的顺序指定列，抽取满足下列全部条件的数据。

- 顾客 ID(`customer_id`)为 "CS018205000001"
- 销售金额(`amount`)在 1,000 以上，或销售数量(`quantity`)在 5 以上

````python
(
df_receipt
.filter(
    pl.col.customer_id
        .eq('CS018205000001')
        .and_(
            pl.col.amount
                .ge(1000)
                .or_(pl.col.quantity >= 5)
        )
)
.select('sales_ymd', 'customer_id', 'product_cd', 'amount', 'quantity')
)
````

## P-007

从收据明细数据(`df_receipt`)中，按销售日(`sales_ymd`)、顾客 ID(`customer_id`)、商品代码(`product_cd`)、销售金额(`amount`)的顺序指定列，抽取满足下列全部条件的数据。

- 顾客 ID(`customer_id`)为 "CS018205000001"
- 销售金额(`amount`)在 1,000 以上、2,000 以下

````python
(
df_receipt
.filter(
    (pl.col.customer_id == 'CS018205000001')
    & pl.col.amount.is_between(1000, 2000)
)
.select('sales_ymd', 'customer_id', 'product_cd', 'amount')
)
````

## P-008

从收据明细数据(`df_receipt`)中，按销售日(`sales_ymd`)、顾客 ID(`customer_id`)、商品代码(`product_cd`)、销售金额(`amount`)的顺序指定列，抽取满足下列全部条件的数据。

- 顾客 ID(`customer_id`)为 "CS018205000001"
- 商品代码(`product_cd`)不是 "P071401019"

````python
(
df_receipt
.filter(
    (pl.col.customer_id == 'CS018205000001')
    & (pl.col.product_cd != 'P071401019')
)
.select('sales_ymd', 'customer_id', 'product_cd', 'amount')
)
````
